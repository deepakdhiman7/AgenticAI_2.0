## Attention vs. Self‑Attention in Transformers  

| Aspect | **Attention (general)** | **Self‑Attention** |
|--------|--------------------------|--------------------|
| **Definition** | A mechanism that relates **one set of elements (the *queries*)** to **another set (the *keys* and *values*)** to compute a weighted sum of the values. | A **special case** of attention where the queries, keys, and values all come from the **same sequence** (the same token embeddings). |
| **Typical use‑cases** | <ul><li>Encoder‑decoder (cross‑attention) – the decoder queries the encoder’s hidden states.</li><li>Memory‑augmented models – queries a separate memory bank.</li></ul> | <ul><li>Encoder self‑attention – each token looks at all other tokens in the same input sentence.</li><li>Decoder self‑attention – each token looks at earlier tokens in the generated output (masked).</li></ul> |
| **Inputs** | <ul><li>**Q** – queries (e.g., decoder hidden states).</li><li>**K**, **V** – keys & values from a *different* source (e.g., encoder hidden states).</li></ul> | <ul><li>**Q**, **K**, **V** all derived from the **same** token matrix **X** (e.g., the embedding matrix of the current layer).</li></ul> |
| **Mathematical form** | \[
\text{Attention}(Q,K,V)=\text{softmax}\!\Big(\frac{QK^{\top}}{\sqrt{d_k}}\Big)V
\] | Same equation, but \(Q=K=V=XW\) (or separate linear projections of the same \(X\)). |
| **Masking** | Usually **no mask**
