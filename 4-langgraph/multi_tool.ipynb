## Learning Multi-tool Calling

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun

load_dotenv()
```

```python
# Setting up environment variables

# print(os.getenv("GROQ_API_KEY"))
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
# loading the model

llm = ChatGroq(model="qwen/qwen3-32b")
```

```python
# testing the model

result = llm.invoke("What is the capital of France?")
```

```python
print(result.content)
```

```python
# Defining tools

# A simple tool to add two numbers
@tool
def add(a: int, b: int) -> int:
    """Add two numbers.
    
    Args:
        a: The first number.
        b: The second number.
        
    Returns:
        The sum of the two numbers.
    """
    return a + b

#Internet search tool
search = DuckDuckGoSearchRun()
```

```python
# Compiling the tools
tools = [add, search]

# adding tools to LLM
llm_with_tools = llm.bind_tools(tools)

# testing the LLM with tools
# result = llm_with_tools.invoke("What is price of INR against USD?")
# result = llm_with_tools.invoke("What will 23434 added to 43234 be?")
```
